```python
# =====================================================
# BANK MARKETING CLASSIFICATION PROJECT
# =====================================================

# =====================================================
# 1. Import Libraries
# =====================================================

import pandas as pd
import numpy as np

from google.colab import files

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.impute import SimpleImputer

from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier

from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import BaggingClassifier
from sklearn.ensemble import AdaBoostClassifier

from sklearn.metrics import accuracy_score
from sklearn.metrics import precision_score
from sklearn.metrics import recall_score
from sklearn.metrics import f1_score
from sklearn.metrics import classification_report

import warnings
warnings.filterwarnings("ignore")

# =====================================================
# 2. Upload Dataset
# =====================================================

print("Select your bank-additional-full.csv file from Downloads")

uploaded = files.upload()

filename = list(uploaded.keys())[0]

# FIXED: Your file is comma-separated, not semicolon-separated
df = pd.read_csv(filename, sep=",")

# =====================================================
# 3. Inspect Dataset
# =====================================================

print("\nFirst 5 Rows")
print(df.head())

print("\nDataset Shape")
print(df.shape)

print("\nDataset Information")
print(df.info())

# =====================================================
# 4. Numerical and Categorical Features
# =====================================================

numerical = df.select_dtypes(include=['int64','float64']).columns

categorical = df.select_dtypes(include=['object']).columns

print("\nNumerical Features")
print(numerical)

print("\nCategorical Features")
print(categorical)

# =====================================================
# 5. Target Variable Distribution
# =====================================================

print("\nTarget Variable Distribution")
print(df["y"].value_counts())

print("\nTarget Percentage")
print(df["y"].value_counts(normalize=True)*100)

# =====================================================
# 6. Handle Unknown Values
# =====================================================

df.replace("unknown", np.nan, inplace=True)

cat_cols = df.select_dtypes(include="object").columns

imputer = SimpleImputer(strategy="most_frequent")

df[cat_cols] = imputer.fit_transform(df[cat_cols])

# =====================================================
# 7. Encode Target Variable
# =====================================================

df["y"] = df["y"].map({"yes":1,"no":0})

# =====================================================
# 8. Encode Categorical Features
# =====================================================

le = LabelEncoder()

for col in cat_cols:
    if col != "y":
        df[col] = le.fit_transform(df[col])

# =====================================================
# 9. Separate Features and Target
# =====================================================

X = df.drop("y", axis=1)

y = df["y"]

# =====================================================
# 10. Feature Scaling
# =====================================================

scaler = StandardScaler()

X = scaler.fit_transform(X)

# =====================================================
# 11. Train Test Split
# =====================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# =====================================================
# 12. Evaluation Function
# =====================================================

def evaluate(model, name):

    model.fit(X_train, y_train)

    prediction = model.predict(X_test)

    print("\n===================================")
    print(name)
    print("===================================")

    print("Accuracy :", accuracy_score(y_test, prediction))
    print("Precision:", precision_score(y_test, prediction))
    print("Recall   :", recall_score(y_test, prediction))
    print("F1 Score :", f1_score(y_test, prediction))

    print("\nClassification Report")
    print(classification_report(y_test, prediction))

    return [
        name,
        accuracy_score(y_test, prediction),
        precision_score(y_test, prediction),
        recall_score(y_test, prediction),
        f1_score(y_test, prediction)
    ]

# =====================================================
# 13. Logistic Regression
# =====================================================

lr = LogisticRegression()

lr_result = evaluate(lr, "Logistic Regression")

# =====================================================
# 14. K-Nearest Neighbors
# =====================================================

knn = KNeighborsClassifier(n_neighbors=5)

knn_result = evaluate(knn, "K-Nearest Neighbors")

# =====================================================
# 15. Support Vector Machine
# =====================================================

svm = SVC()

svm_result = evaluate(svm, "Support Vector Machine")

# =====================================================
# 16. Decision Tree
# =====================================================

dt = DecisionTreeClassifier(random_state=42)

dt_result = evaluate(dt, "Decision Tree")

# =====================================================
# 17. Random Forest
# =====================================================

rf = RandomForestClassifier(random_state=42)

rf_result = evaluate(rf, "Random Forest")

# =====================================================
# 18. Bagging Classifier
# =====================================================

bag = BaggingClassifier(random_state=42)

bag_result = evaluate(bag, "Bagging Classifier")

# =====================================================
# 19. AdaBoost Classifier
# =====================================================

ada = AdaBoostClassifier(random_state=42)

ada_result = evaluate(ada, "AdaBoost")

# =====================================================
# 20. Compare All Models
# =====================================================

comparison = pd.DataFrame(
    [
        lr_result,
        knn_result,
        svm_result,
        dt_result,
        rf_result,
        bag_result,
        ada_result
    ],
    columns=[
        "Model",
        "Accuracy",
        "Precision",
        "Recall",
        "F1 Score"
    ]
)

print("\n==============================")
print("MODEL COMPARISON")
print("==============================")

print(comparison.sort_values(by="Accuracy", ascending=False))

# =====================================================
# 21. Feature Importance (Random Forest)
# =====================================================

rf.fit(X_train, y_train)

importance = pd.Series(
    rf.feature_importances_,
    index=df.drop("y", axis=1).columns
)

importance = importance.sort_values(ascending=False)

print("\nTop 10 Important Features")

print(importance.head(10))

# =====================================================
# 22. Best Model
# =====================================================

best = comparison.sort_values(by="Accuracy", ascending=False)

print("\nBest Performing Model")

print(best.head(1))
```
